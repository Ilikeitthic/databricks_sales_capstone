```sql
-- Data Quality trend: flag counts and % per pipeline run
-- Visualise as a stacked bar chart: x = run_timestamp, y = count, colour = quality_flag
SELECT
    run_timestamp,
    quality_flag,
    count,
    ROUND(100.0 * count / SUM(count) OVER (PARTITION BY run_timestamp), 2) AS pct
FROM workspace.default.capstone_dq_summary
ORDER BY run_timestamp DESC, count DESC;
```
